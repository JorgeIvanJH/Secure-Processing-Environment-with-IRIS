```python
import iris.dbapi
import pandas as pd

query = "SELECT TOP 5 * FROM MockPackage.NoShowsAppointments"
```

```python
# Yomna can read the table.
connection_string = "localhost:1972/USER"
connection = iris.dbapi.connect(connection_string, "Yomna", "demo")
cursor = connection.cursor()
cursor.execute(query)
yomna_df = pd.DataFrame(cursor.fetchall(), columns=[column[0] for column in cursor.description])
connection.close()
yomna_df
```

```python
# Jorge can connect, but cannot read the table.
connection_string = "localhost:1972/USER"
try:
    connection = iris.dbapi.connect(connection_string, "Jorge", "demo")
    cursor = connection.cursor()
    cursor.execute(query)
    jorge_df = pd.DataFrame(cursor.fetchall(), columns=[column[0] for column in cursor.description])
    jorge_df
except Exception as error:
    print(f"Access denied as expected: {error}")
finally:
    connection.close()
```
